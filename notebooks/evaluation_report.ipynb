# Evaluation Report
This notebook generates a combined PDF with two sections: Chatbot NLP and Multilingual Language Selector.

```python
import os
import pandas as pd
from IPython.display import display
from PIL import Image as PILImage

base_images = 'docs/images'
out_pdf = 'docs/evaluation_report.pdf'
os.makedirs(os.path.dirname(out_pdf), exist_ok=True)
```

## Section 1: Chatbot NLP
Load the intent evaluation tables and charts for chatbot understanding metrics.

```python
intent_csv = os.path.join(base_images, 'per_intent_metrics.csv')
if os.path.exists(intent_csv):
    display(pd.read_csv(intent_csv))
else:
    print('Missing intent metrics CSV:', intent_csv)
```

```python
intent_image = os.path.join(base_images, 'intent_confusion_matrix.png')
if os.path.exists(intent_image):
    display(PILImage.open(intent_image))
else:
    print('Missing intent confusion matrix image:', intent_image)
```

## Section 2: Multilingual Language Selector
Load language selector evaluation tables and charts for multilingual detection metrics.

```python
lang_csv = os.path.join(base_images, 'per_language_accuracy.csv')
if os.path.exists(lang_csv):
    display(pd.read_csv(lang_csv))
else:
    print('Missing language metrics CSV:', lang_csv)
```

```python
lang_images = [
    os.path.join(base_images, 'accuracy_bar.png'),
    os.path.join(base_images, 'confusion_matrix.png')
]
for path in lang_images:
    if os.path.exists(path):
        display(PILImage.open(path))
    else:
        print('Missing language selector image:', path)
```

## Supported Languages
List all languages covered by the multilingual evaluation.

```python
lang_csv = os.path.join(base_images, 'per_language_accuracy.csv')
if os.path.exists(lang_csv):
    df = pd.read_csv(lang_csv)
    if 'language' in df.columns:
        print('Supported languages:', ', '.join(df['language'].astype(str).tolist()))
    else:
        print('Supported languages information not available in CSV.')
else:
    print('Missing language metrics CSV:', lang_csv)
```
